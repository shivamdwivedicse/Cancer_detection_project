```python
df = df.drop(columns=['id',"Unnamed: 32"])
```

```python
df["diagnosis"] = df["diagnosis"].map({"M":1 , "B":0})
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop(columns=['diagnosis'])
y = df['diagnosis']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler 
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,classification_report
```

```python
#by using grid search cv we get c =0.5 is beat hyperparameter :-
# you can try it from running this cell by cell:-

#gridsearchcv predictions and score for logestic regression model:-


# 1st cell:-
# from sklearn.model_selection import GridSearchCV
# params = {
#     'C':[0.1,0.5,1.5,10]
# }
# grid = GridSearchCV(LogisticRegression(max_iter=2000),params,cv=5)
# grid.fit(x_train , y_train)

# 2nd cell:-
# y_pred_grid = grid.predict(x_test)
# y_pred_grid

# 3rd cell:-
# accuracy_score(y_test , y_pred_grid)

#4th cell :-
# cnn_grid = confusion_matrix(y_test , y_pred_new)
# cnn_grid

#5th cell:-
# scores_grid = cross_val_score(grid , x , y , cv = 5)
# scores_grid

#6th cell:-
# scores_grid.mean()

#7th cell:-
#  grid.best_params_

model = LogisticRegression(C=0.5,max_iter=2000 , class_weight="balanced")

model.fit(x_train , y_train)  #kepping threshold = 0.5 
```

```python
y_pred = model.predict(x_test)
```

```python
accuracy_score(y_test , y_pred)
```

```python
print(classification_report(y_test , y_pred))
```
